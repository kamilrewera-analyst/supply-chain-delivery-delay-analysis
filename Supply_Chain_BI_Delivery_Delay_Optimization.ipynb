## **1. Data Dictionary**

### **Metrics Glossary**
- **Total Orders:** Total number of active orders in the analyzed scope (excluding *Shipping canceled* via an active context filter).
- **Late Orders:** Total number of delayed orders within a given dimension/segment.
- **Late Delivery Rate:** Share of late orders out of *Total Orders* (`Late Orders / Total Orders`).
- **Average Delay in Days:** Average delay calculated across late orders (*Average Delay of Late Orders in Days*).
- **Delayed Sales:** Net sales value (*Delayed Sales (Net)*) associated with late orders.

### **1. Duplicate Columns Check & Shipping Data Consistency Check**

```python
# duplicated cols
duplicates = []

for i in range(data.shape[1]):
    for j in range(i + 1, data.shape[1]):

        if data.iloc[:, i].equals(data.iloc[:, j]):
            duplicates.append({
                'Original': data.columns[i],
                'Duplicate': data.columns[j]
            })

duplicates_df = pd.DataFrame(duplicates)

print(f"--- Duplicated cols ---")
print(duplicates_df)

# shipping consistency within Order ID
consistency_check = (
    data.groupby('Order Id')
    .agg({
        'Shipping Mode': 'nunique',
        'Delivery Status': 'nunique',
        'Days for shipping (real)': 'nunique',
        'Days for shipment (scheduled)': 'nunique',
        'order date (DateOrders)': 'nunique',
        'shipping date (DateOrders)': 'nunique'
    })
    .reset_index()
)

print(f"\n--- Shipping Inconsistent within Order ID---")
inconsistent_orders = consistency_check[
    (consistency_check.drop(columns='Order Id') > 1).any(axis=1)
]

print(f"Number of inconsistent orders: {inconsistent_orders.shape[0]}")
```

**Shipping Consistency Check**

Since the dataset is structured at the line-item level (`Order Item Id`), we verified whether all items belonging to the same `Order Id` share identical shipping and delivery attributes (such as `Shipping Mode`, `Delivery Status`, scheduled/real shipping days, and order/shipping timestamps).
